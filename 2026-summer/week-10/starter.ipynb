## 1. Get the data

The transactions ship as a single `transactions.zip`. The cell below downloads it and unzips it into a `transactions/` folder. The challenge is on purpose: the files are zipped, so you have to unzip before you can read them.

```python
import urllib.request, zipfile, os

BASE = "https://raw.githubusercontent.com/sean-mccaman/acctg5150-090/main/2026-summer/week-10"

urllib.request.urlretrieve(f"{BASE}/transactions.zip", "transactions.zip")
with zipfile.ZipFile("transactions.zip") as z:
    z.extractall(".")

files = sorted(os.listdir("transactions"))
print(f"Unzipped {len(files)} files. First few: {files[:3]}")
```

## 2. Read the whole folder into one DataFrame

There is no single combined file. You read every daily CSV and stack them. `glob` lists the files that match a pattern; `pd.concat` stacks them into one DataFrame. This is the pattern for "read a folder of files".

```python
import glob
import pandas as pd

paths = sorted(glob.glob("transactions/txn_*.csv"))
print(f"Found {len(paths)} daily files")

df = pd.concat((pd.read_csv(p) for p in paths), ignore_index=True)
print("Combined shape:", df.shape)
df.head()
```
